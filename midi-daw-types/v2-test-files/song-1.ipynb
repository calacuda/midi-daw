```python
from midi_daw import MidiChannel, NoteLen, v2
from midi_daw.v2 import lfo, main
from midi_daw.v2.note_lens import *
```

```python
song = v2.MidiDaw("Vital", MidiChannel.Ch1, virt=True)
```

```python
automa = v2.MidiDaw("Vital", MidiChannel.Ch1, virt=True)
```

```python
@song.register
def lead(m):
    m.play("g5", qn())
    m.play("c5", en())
    m.play("d#5", sn())
    m.play("f5", sn())
    m.rest(sn())
    m.play("a#4", qn())
    m.play("g4", en())
    m.play("a#4", sn())
    m.play("c5", sn())
    m.rest(sn())
    m.play("g4", qn())
    m.play("c4", en())
    m.play("d#4", sn())
    m.play("f4", sn())
    m.rest(sn())
```

```python
l_jh = lead(loops=25, block=False)
```

```python
song.stop(l_jh)
```

```python
bass = v2.MidiDaw("Bass", MidiChannel.Ch1, virt=True)
```

```python
@bass.register
def bass_f(m):
    m.play("g3", qn())
    m.play("c3", en())
    m.play("d#3", sn())
    m.play("f3", sn())
    m.rest(sn())
    m.play("a#2", qn())
    m.play("g2", en())
    m.play("a#2", sn())
    m.play("c2", sn())
    m.rest(sn())
    m.play("g2", qn())
    m.play("c2", en())
    m.play("d#2", sn())
    m.play("f3", sn())
    m.rest(sn())
```

```python
b_jh = bass_f(loops=25, block=False)
```

```python
bass.stop(b_jh)
```
